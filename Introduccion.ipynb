# Python y políticas públicas

Bienvenidos y bienvenidas! Este es un curso introductorio de python para estudiantes de políticas públicas o interesados/as en el análisis de bases de datos e información proveniente o para el sector público. 

El uso de evidencia para la toma de decisiones en el sector público es una de los grandes desafíos para lograr tener administraciones eficaces y efectivas. En la actualidad, la enorme disponibilidad de datos y la creciente creación de nueva información requieren que las personas profesionales de la políticas públicas tengan una fuerte formación en análisis, manipulación y visualización de datos. 

Python es un lenguaje que se destaca por su versatilidad ya que permite realizar todo tipo de análisis estadísticos, páginas webs, modelos de inteligencia artificial, manipulación de bases georeferenciadas, visualización en dashboards y mucho, mucho más. Es uno de los lenguajes de programación mas utilizados en la actualidad y su gran comunidad de usuarios y usuarias hace que las potencialidades de Python crezcan día a día de forma exponencial. Además, Python es _open source_, lo cual lo hace muy participativo y colaborativo. No te quedes afuera de este barco!
