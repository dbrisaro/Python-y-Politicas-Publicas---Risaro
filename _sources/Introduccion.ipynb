# Python y políticas públicas

Bienvenidos y bienvenidas! Este es un curso introductorio de Python para estudiantes de políticas públicas o interesados/as en el análisis de bases de datos e información proveniente o para el sector público.

El uso de evidencia para la toma de decisiones en el sector público es uno de los grandes desafíos para lograr administraciones eficaces y efectivas. En la actualidad, dada la enorme disponibilidad de datos, las personas profesionales de las políticas públicas requieren conocer las bases del análisis, la manipulación y la visualización de datos, para generar evidencias que sustenten la toma de decisiones.

Python es un lenguaje que se destaca por su versatilidad: permite realizar todo tipo de análisis estadísticos, implementar algoritmos de aprendizaje automático, manipular bases georreferenciadas, visualizar información en dashboards, crear páginas web, y mucho más. Es uno de los lenguajes de programación más utilizados en la actualidad y su gran comunidad hace que sus potencialidades crezcan día a día en forma exponencial. Además, Python es _open source_, lo cual lo hace participativo y colaborativo. ¡No te quedes afuera de este barco!

## Estructura del curso

Este curso se divide en **10 clases** con contenido teórico y práctico, y culmina con el desarrollo de un **proyecto final individual** en formato de _policy brief_ con datos.

---

### Clase 1 — Introducción a Python
- ¿Qué es Python? Intérpretes disponibles. Introducción a Anaconda.
- Primeros pasos en Jupyter Notebooks.
- Tipos de datos y operaciones: enteros, flotantes, strings, booleanos.
- Estructuras de control: `if`, `while`, `for`.

### Clase 2 — Listas, funciones y numpy
- Introducción a `numpy`: arrays uni y multidimensionales, operaciones vectorizadas.
- Creación y operaciones sobre listas.
- Diccionarios.
- Definición y uso de funciones.
- _Ejercicio integrador_: combinar funciones, numpy, listas y un `for`.

### Clase 3 — Matplotlib y visualización
- Buenas prácticas en visualización de datos.
- Tipos de gráficos: line plots, scatter plots, bar charts, histogramas, heatmaps, series temporales.
- ¿Qué gráfico elegir según el tipo de variable?
- Personalización: títulos, etiquetas, colores, estilos.

### Clase 4 — Pandas I: lectura y exploración de datos
- Lectura de archivos: `.csv`, `.xlsx`, `.sav` y otros formatos.
- Tipos de datos de pandas: `DataFrame` y `Series`.
- Exploración inicial: `.head()`, `.info()`, `.describe()`, `.dtypes`.
- Lectura de datos de fuentes públicas (datos.gob.ar, portales provinciales).

### Clase 5 — Pandas II: operaciones sobre DataFrames
- Selección de datos: indexing, selecting, slicing.
- Merge, join, concatenate.
- Reshaping y pivot tables.
- GroupBy y agregaciones.

### Clase 6 — Data cleansing
- Valores faltantes: detección, eliminación e imputación.
- Duplicados y inconsistencias.
- Conversión de tipos de datos.
- Normalización y estandarización de variables.
- _Ejercicio integrador_: limpiar un dataset real del sector público.

### Clase 7 — Pandas + Matplotlib: integración
- Exploración visual de DataFrames: distribuciones, outliers, series temporales.
- Subplots y figuras con múltiples paneles.
- Gráficos directamente desde pandas.
- Exportar gráficos con calidad de publicación.

### Clase 8 — Estadística aplicada con Python
- Estadística descriptiva con `pandas` y `scipy.stats`.
- Correlaciones e interpretación.
- Regresión lineal con `statsmodels` (OLS): especificación, estimación, interpretación de coeficientes.
- Tablas de resultados listas para incluir en un _policy brief_.

### Clase 9 — Datos geoespaciales
- ¿Qué son los datos geoespaciales? Formatos: shapefiles, GeoJSON.
- Introducción a `geopandas`.
- Mapas coropléticos: visualizar indicadores por provincia o municipio.
- Integración con datos del sector público argentino (INDEC, datos.gob.ar).

### Clase 10 — APIs, datos abiertos y proyecto final
- ¿Qué es una API? Cómo consumir datos de APIs públicas con `requests`.
- APIs relevantes para políticas públicas: datos.gob.ar, World Bank, CEPAL.
- Estructura del proyecto final: cómo armar un _policy brief_ con datos en Python.
- Taller de cierre: presentación de avances y consultas.

---

## Proyecto final

El proyecto final consiste en la elaboración de un **policy brief con datos** usando Python. Cada estudiante elige un tema de política pública de su interés, identifica fuentes de datos abiertas relevantes, realiza un análisis en Python (limpieza, exploración, visualización y al menos una técnica estadística), y presenta sus hallazgos en formato de policy brief (2–4 páginas).

**Criterios de evaluación:**
- Calidad y pertinencia del análisis de datos.
- Claridad de las visualizaciones.
- Solidez del argumento de política pública.
- Reproducibilidad del código (Jupyter Notebook entregado).

## FAQ

**¿Necesito tener experiencia previa para tomar este curso?**  
No. Aprenderemos a programar en Python desde cero. Si tenés experiencia previa, siempre viene bien refrescar conceptos.

**¿Me sirve saber R o Stata?**  
Sí. Tener experiencia de programación en otros lenguajes o herramientas estadísticas es un plus, aunque en este curso nos enfocaremos en Python.

**¿Qué herramientas necesito instalar?**  
Recomendamos instalar [Anaconda](https://www.anaconda.com/), que incluye Python, Jupyter Notebooks y las principales librerías que usaremos. Las instrucciones de instalación se proveen en la Clase 1.

**¿Qué datos vamos a usar?**  
Trabajaremos principalmente con datos abiertos del sector público argentino (datos.gob.ar, INDEC, portales provinciales) y fuentes internacionales (World Bank, CEPAL). Todos los datasets son de libre acceso.
